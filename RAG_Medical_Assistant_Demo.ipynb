# 🧪 RAG Demo for a Medical Assistant
This notebook shows a simple Retrieval-Augmented Generation (RAG) pipeline using LangChain, ChromaDB, and OpenAI for a medical Q&A assistant.

```python
# Step 1: Import necessary libraries
from langchain.embeddings import OpenAIEmbeddings
from langchain.vectorstores import Chroma
from langchain.chat_models import ChatOpenAI
from langchain.chains import RetrievalQA
import os
```

```python
# Step 2: Set up your OpenAI API key
os.environ['OPENAI_API_KEY'] = 'your-api-key-here'
```

```python
# Step 3: Load medical documents
medical_notes = [
    "Hypertension is a chronic condition that can lead to heart disease.",
    "Patients with diabetes should monitor their blood sugar levels regularly.",
    "Amoxicillin is commonly prescribed for bacterial infections.",
    "Chest pain accompanied by shortness of breath may indicate a cardiac issue."
]
```

```python
# Step 4: Embed and store in Chroma vector DB
embeddings = OpenAIEmbeddings()
db = Chroma.from_texts(medical_notes, embedding=embeddings, collection_name='medical-assistant')
```

```python
# Step 5: Create retriever and QA chain
retriever = db.as_retriever(search_kwargs={"k": 2})
qa_chain = RetrievalQA.from_chain_type(llm=ChatOpenAI(), retriever=retriever)
```

```python
# Step 6: Ask a medical question
response = qa_chain.run("What medication is commonly used for bacterial infections?")
print(response)
```
